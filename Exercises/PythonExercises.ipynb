## Exercise 12

Given the mRNA sequence:

```text
UAUAAACGAUACCAUUACUAUGACCAUGGG
```

It consists of 10 codons. Tyrosine is encoded by `UAU` and `UAC`. Find the codon positions encoding tyrosine.

Hint:

```python
for index, value in enumerate(my_list, start=1):
    ...
```


````python
# Source - https://stackoverflow.com/a/44696587
# Posted by Tim Diels
# Retrieved 2026-10-08, License - CC BY-SA 3.0

# the answer
from more_itertools import sliced
seq = 'UAUAAACGAUACCAUUACUAUGACCAUGGG'
codons = list(sliced(seq, 3))

for index, value in enumerate(codons, start=1):
    if value == 'UAU' or value == 'UAC':
        print(index)

print("With sliding window - reading all possible OPF:")
tyrosine = ['UAU','UAC']

pos = 0
codons = []
while pos <= len(seq):
    codons.append(seq[pos:pos+3])
    pos = pos +1

for index, codon in enumerate(codons, start = 1):
    if codon in tyrosine:
        print(index, codon)
````

## Exercise 4

Parse `paxillus.fna` and calculate GC-content for each sequence. Write a new file with the GC percentage appended to the header.

Example:

```text
>id_for_my_entry additional_info
GGCCAA
```

becomes:

```text
> id_for_my_entry additional_info GC:66.7%
GGCCAA
```

The script should accept input and output files as arguments.


````python
# the answer
import sys

def add_gc_content(input_file, output_file):
    with open(input_file, "r") as infile, open(output_file, "w") as outfile:

        header = None
        sequence_parts = []

        for line in infile:
            line = line.strip()

            if line.startswith(">"):
                # Write the previous sequence before starting a new one
                if header is not None:
                    sequence = "".join(sequence_parts).upper()
                    gc = (
                        (sequence.count("G") + sequence.count("C"))
                        / len(sequence) * 100
                        if sequence else 0
                    )
                    outfile.write(f"> {header} GC:{gc:.1f}%\n")
                    outfile.write(sequence + "\n")

                header = line[1:]
                sequence_parts = []
            else:
                sequence_parts.append(line)

        # Don't forget the final sequence
        if header is not None:
            sequence = "".join(sequence_parts).upper()
            gc = (
                (sequence.count("G") + sequence.count("C"))
                / len(sequence) * 100
                if sequence else 0
            )
            outfile.write(f"> {header} GC:{gc:.1f}%\n")
            outfile.write(sequence + "\n")

    print("b) GC content written to", output_file)


# a) Set the input and output files
input_file = "Data/paxillus.fna" 
output_file = "Data/paxillus_gc.fna"

print("a) Input file:", input_file)
print("   Output file:", output_file)

# b) Process the FASTA file
add_gc_content(input_file, output_file)
````

## Exercise 6

Write a program converting FASTQ to FASTA. Make sure you understand the file format; modulo operations can be useful.


````python
# the answer
import sys

def fastq_to_fasta(input_file, output_file):
    with open(input_file, "r") as infile, \
         open(output_file, "w") as outfile:

        for line_number, line in enumerate(infile, start=1):
            line = line.rstrip("\r\n")

            # Line 1 of each four-line FASTQ record: header
            if line_number % 4 == 1:
                if not line.startswith("@"):
                    raise ValueError(
                        f"Expected FASTQ header at line {line_number}"
                    )
                outfile.write(">" + line[1:] + "\n")

            # Line 2: nucleotide sequence
            elif line_number % 4 == 2:
                outfile.write(line + "\n")

            # Line 3: plus separator; skip it
            elif line_number % 4 == 3:
                if not line.startswith("+"):
                    raise ValueError(
                        f"Expected '+' separator at line {line_number}"
                    )

            # Line 4: quality scores; skip them
            elif line_number % 4 == 0:
                pass

        if line_number % 4 != 0:
            raise ValueError("Incomplete FASTQ record at end of file")

    print("a) FASTQ file:", input_file)
    print("b) FASTA file created:", output_file)


# Notebook test:
input_file = "Data/fungus.fastq"       # TODO: Change to your FASTQ filename
output_file = "Data/fungus.fasta"

print("Converting FASTQ to FASTA:")
fastq_to_fasta(input_file, output_file)
````
